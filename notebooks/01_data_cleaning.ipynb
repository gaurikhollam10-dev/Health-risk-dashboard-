# 01 — Data Cleaning

Cleans and validates the healthcare CSV, removes duplicates, handles missing values and obvious invalid ranges, and saves `data/cleaned_data.csv`.

```python
from pathlib import Path
import sys
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / 'data').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

from src.preprocessing import clean_dataframe, validate_columns, canonicalize_columns

raw = pd.read_csv(ROOT / 'data' / 'healthcare_risk.csv')
print('Raw shape:', raw.shape)
print('Columns:', list(raw.columns))
```

```python
raw = canonicalize_columns(raw)
valid, missing = validate_columns(raw, require_target=True)
if not valid:
    raise ValueError(f'Missing required columns: {missing}')

cleaned = clean_dataframe(raw, require_target=True)
print('Cleaned shape:', cleaned.shape)
print('Duplicates after cleaning:', cleaned.duplicated().sum())
print('\nMissing values:')
print(cleaned.isna().sum())
```

```python
cleaned.to_csv(ROOT / 'data' / 'cleaned_data.csv', index=False)
cleaned.head()
```
